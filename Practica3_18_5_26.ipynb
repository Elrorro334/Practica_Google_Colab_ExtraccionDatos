```python
import pandas as pd
from google.colab import files
import matplotlib.pyplot as plt
```

subir dataset

```python
uploaded = files.upload()
df = pd.read_csv('ventasnvo.csv')
```

procesamiento de datos mostrar las primeras filas

```python
print("\n Encabezado del dataset")
print(df.head())
```

Informacion general del dataset

```python
print("\n Informacion del dataset")
print(df.info())
print("\n Descripcion del dataset")
print(df.describe())
```

Conteo de categorias

```python
print("\n Cantidad por categoria")
print(df['Categoria'].value_counts())

print("\n Cantidad por pago")
print(df['MetodoPago'].value_counts())

print("\n Producto mas Caro")
print(df.loc[df['Precio'].idxmax()])
```

```python
print("\n Promedio y Suma")
print(df['Precio'].mean())
print(df['Cantidad'].sum())
```

```python
print("\n Graficas de Categorias y Cantidad de Ventas")
df['Categoria'].value_counts().plot(kind='bar')
```

```python
print("\n Graficas de Categorias y Porcentaje de Ventas Grafica de Pastel")
df['Categoria'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Porcentajes de Categorias')
plt.ylabel('')
plt.show()
```

```python
print("\n Graficas de Precio en histograma")
df['Precio'].plot(kind='hist')
plt.title('Distribucion de precios')
plt.xlabel('')
plt.show()
```

```python
df.plot(kind='scatter', x='Precio', y='Cantidad')
plt.title('Precio vs Cantidad')
plt.xlabel('Precio')
plt.ylabel('Cantidad')
```
